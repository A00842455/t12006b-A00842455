# Activity 3: Colab–VS Code Bridge

## Research Question
What is the relationship between study hours and exam scores?

## Data Provenance
This notebook uses synthetic data created with Python. No personal or sensitive information is included.

## Expected Output
Analyze the relationship between study hours and exam scores using a table, summary statistics, and a visualization.

```python
import sys
import pandas as pd
import matplotlib.pyplot as plt

print("Python version:", sys.version)
print("Pandas version:", pd.__version__)
print("Matplotlib is ready!")
```

```python
# Create synthetic student data

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [2, 3, 4, 5, 6, 8],
    "Exam_Score": [60, 65, 70, 78, 85, 95]
}

df = pd.DataFrame(data)

print("Synthetic Student Data:")
print(df)
```

```python
# Analyze the synthetic data

print("Average Study Hours:", df["Study_Hours"].mean())
print("Average Exam Score:", df["Exam_Score"].mean())

correlation = df["Study_Hours"].corr(df["Exam_Score"])
print("Correlation:", round(correlation, 2))

# Create visualization
plt.scatter(df["Study_Hours"], df["Exam_Score"])

plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)

plt.show()
```

## Conclusion

The analysis shows a positive relationship between study hours and exam scores in the synthetic dataset. Students who study more hours tend to achieve higher scores.
